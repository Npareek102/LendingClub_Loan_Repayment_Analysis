# Four Categories of Columns in Credit Risk

1. Borrower Characteristics : These describe the person taking the loan. eg: annual_inc, emp_length, home_owernship, verification_status, addr_state
2. Loan Characteristics : These describe the loan contract. eg: loan_amnt, term, int_rate, grade, sub_grade, purpose, installment
3. Credit History Variables : These come from the borrowers credit bureau report. eg: dti, delinq_2yrs, open_acc, revol_bal, public_rec_bankruptcies. These measure debt levels, credit utilization, past delinquencies.
4. Post Loan/Repayment Variables : These describe what happened after loan started. eg: total_pymnt, recoveries, last_pymnt_d, total_rec_prncp, collection_recovery_fee.

This classification matters in making the decision of which variables to keep and which to remove. 

```python
df.columns.to_list()
```

# Exploratory Risk Analysis

```python
#grade is Lending Club's internal risk rating
df['grade'].value_counts()
```

```python
df.groupby('grade')['default'].mean().sort_index().plot(kind='bar')
plt.ylabel("Default Rate")
plt.title("Default Rate by Loan Grade")
plt.show()
```

Borrowers classified as high risk (F-G) actually actually default much more frequently

```python
df['grade'].value_counts().sort_index()
```

Loan grades A-C represents majority of the borrowers  
Although the number of borrower's from grade G are lower still the number is usable.  
This also highlights that the platform approves far more lower risk borrowers than high risk ones.

```python
df.groupby('purpose')['default'].mean().sort_values(ascending=False)
```

```python
df.groupby('annual_inc')['default'].mean().sort_values(ascending=False)
```

```python
df['purpose'].value_counts()
```

Loan purpose significantly influences default risk.

Small business loans exhibit the highest default rate (~30%),
while car and credit card loans show comparatively lower risk.

Debt consolidation accounts for the majority of loans in the
portfolio (~56%) and therefore contributes the largest share
of total defaults despite having only moderate default rates.

```python
df['income_group'] = pd.cut(
    df['annual_inc'],
    bins=[0, 40000, 80000, 120000, 200000, 1000000],
    labels=['Low', 'Lower-Mid', 'Mid', 'Upper-mid', 'High'],
)

df.groupby('income_group', observed=False)['default'].mean().sort_values(ascending=False)
```

Income alone should not be used to reject loans, but it is useful when combined with other variables like:
DTI (debt-to-income ratio)
credit history
loan grade

```python
df.groupby('emp_length')['default'].mean().sort_values()
```

```python
df['emp_length'].value_counts()
```

Unlike income or grade, employment length lacks a strong monotonic pattern. Default risk doesn't consistently increase or decrease with employment length. This happens because employement length alone doesn't represent financial stability. So it is a weak standalone predictor.  
This indicates lender rely more on credit score, income, loan grade.

Now we measure Debt to income ratio (DTI), it measures how much debt a borrower already has relative to income. Higher DTI usually means higher risk.

```python
df.groupby(pd.cut(df['dti'], bins=[0,10,20,30,40,100]), observed=False)['default'].mean()
```

```python
df.groupby(pd.cut(df['dti'], bins=[0,10,20,30,40,100]), observed=False)['default'].count()
```

Default risk substantially increases as debt to income ratio rises, reaching its highest levels among borrowers with DTI 30-40%. It's sudden decrease in the >40 category is likely due to very small number of observations in this range.  
This also may be a sign of underwriting policy wherein lender typically avoid approving borrowers with very high DTI, maybe the platform filters out many application cases above ~40 DTI, leaving only a few exception cases.
So DTI is a strong indicator of default risk.

```python
df.groupby('home_ownership')['default'].mean()
```

```python
df['home_ownership'].value_counts()
```

Renters often have less financial stability so they are less capable to absorb financial shocks.  
People with Mortage often had to pass strict credit checks to obtain a home loan and they already have an established credit history.  
This suggests that housing stability correlates with lower credit risk.

```python
df.groupby(pd.cut(df['fico_score'], bins=[600,650,700,750,800,850]), observed=False)['default'].mean()
```

One of the strongest and most expected patterns in credit risk  
High FICO score --> Lower credit risk  
Borrower's with lower credit score default about 4x more often then borrowers with excellent score.  
Range 600-650 might be NAN bcs no loan might be falling into the range.

```python
df.groupby(pd.cut(df['fico_score'], bins=[600,650,700,750,800,850]),observed=False)['default'].count()
```

As we discussed the chances of no loan falling in 600-650 fico score range, can be seen in this output.  
This is the same reason as of DTI wherein many platforms automatically reject such applicants based on their internal policies, in this case of credit scores.

A lender might use this insights for making business decisions wherein:
    - Offer lower interest rates for applicants having high FICO score.
    - Set stricter approval rules for low FICO borrowers.
